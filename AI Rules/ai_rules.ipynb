## 📝 4. Markdown Formatlaşdırma və Sənədləşdirmə Qaydaları

Sənədləşdirmə və Notebook daxilində vizual oxunaqlılığı təmin etmək üçün aşağıdakı Markdown qaydalarına riayət olunmalıdır:

### 4.1. Başlıq İyerarxiyası (Headings)
- Hər bir sənəddə yalnız bir ədəd Ana Başlıq (`# H1`) olmalıdır.
- Bölmələr üçün `#`, `##`, `###` ardıcıllığı pozulmamalıdır.

### 4.2. Kod Blokları (Code Formatting)
- Kod nümunələri yazılarkən sintaksis işıqlandırılması üçün sintaksis dili mütləq göstərilməlidir (`python`, `bash`, `json`, `sql` və s.):
  ```python
  def hello_world():

  ```

### 4.3. Mətn Fərqləndirməsi və Siyahılar
- Mühüm anlayışlar **qalın (bold)** və ya *kursiv (italic)* olaraq vurğulanmalıdır.
- Elementlər ardıcıllıq tələb edirsə nömrələnmiş (`1.`, `2.`), əks halda maddəli (`-` və ya `*`) siyahı istifadə olunmalıdır.

### 4.4. Cədvəllər (Tables)
- Strukturlaşdırılmış data və müqayisələr üçün Markdown cədvəl formatından istifadə olunmalıdır:
  | Parametr | Təsviri | Nümunə |
  | :--- | :--- | :--- |
  | `lr` | Öyrənmə dərəcəsi | `0.001` |
  | `epochs` | İterasiya sayı | `50` |

### 4.5. Qeydlər və Xəbərdarlıqlar (Quotes / Alerts)

## 📊 6. Məlumatın Vizualizasiyası və Matplotlib Qaydaları (Data Visualization)

Data analizi və vizualizasiya tapşırıqlarında riayət olunmalı əsas prinsiplər:

### 6.1. Xətti Qrafiklər (Line Plots)
- Trendlərin göstərilməsində `plt.plot()` istifadə olunur.
- Nöqtələr mütləq `marker` (məs. `'o'`) və `linestyle` (məs. `'--'`) parametrləri ilə fərqləndirilməlidir.
- Şəbəkə xətləri (`plt.grid(True)`) və uyğun `figsize` tənzimlənməlidir.
- Hər bir nöqtənin dəqiq qiyməti `plt.text()` ilə nöqtənin üstündə göstərilməlidir.

### 6.2. Sütunlu Qrafiklər (Bar & Horizontal Bar Charts)
- Kateqoriyalar üzrə müqayisə üçün `plt.bar()` və ya `plt.barh()` istifadə olunur.
- Ən yüksək göstəriciyə malik kateqoriya vizual olaraq (fərqli rəng və ya kontur ilə) vurğulanmalıdır.
- Hər barın sonunda göstərici dəyəri `plt.text()` ilə qeyd olunmalıdır.
- Y oxunun yuxarıdan aşağı düzülüşü üçün `plt.gca().invert_yaxis()` istifadə edilə bilər.

### 6.3. Dağılma Qrafikləri və Xüsusi Oxlar (Scatter & Custom Axes)
- Strukturlaşdırılmış Figure yaradılması üçün `fig = plt.figure()` və `fig.add_axes()` metodlarından istifadə olunmalıdır.
- Çoxölçülü məlumatlar rəng şkalası (`cmap`) və nöqtə ölçüsü (`s`) ilə əks etdirilməli, `plt.colorbar()` əlavə edilməlidir.
- Xüsusi/maksimum göstəricilər `ax.annotate()` vasitəsilə ox və mətnlə fərqləndirilməlidir.
- Şəkil daxilində ikinci köməkçi qrafik üçün Inset Axes (`fig.add_axes(...)`) metodundan istifadə edilə bilər.
