## 📌 1. Kodlaşdırma və Əməkdaşlıq Qaydaları

1. **Təmiz və Strukturlaşdırılmış Kod (PEP 8):** Bütün Python kodları oxunaqlı olmalı, dəyişən adları aydın və təbii seçilməlidir.
2. **Təbii Kod Üslubu (Human-like Code Style):** Kod yazılarkən AI şablonlarından, həddindən artıq bəzəkli süni emoji/bannerlərdən qaçınmalı, adiliyi və sadəliyi qorumaq lazımdır.
3. **Modulluq:** Böyük kod blokları funksiyalara və modullara bölünməlidir.
4. **Sənədləşdirmə və Şərhlər:** Kommentlər sadə, aydın və təbii dildə yazılmalıdır.

## 🔒 2. Məlumat Təhlükəsizliyi və Məxfilik (Data Privacy)

1. **Məxfilik:** Şəxsi məlumatlar (API Key-lər, parol, istifadəçi məlumatları) kod daxilində saxlanılmamalı, `.env` fayllarında qorunmalıdır.
2. **Təmiz Repozitoriya:** Model çəkiləri (weights) və böyük dataset-lər git-ə yüklənməməli, `.gitignore` faylına əlavə edilməlidir.

## 🎯 3. Prompt Mühəndisliyi (Prompt Engineering Rules)

1. **Kontekst Təminatı:** Modellərə sorğu verərkən aydın rol, kontekst və gözlənilən nəticə formatı (JSON, Markdown və s.) göstərilməlidir.
2. **Yoxlama və Test:** Modelin verdiyi nəticələr həmişə insan müayinəsindən (Human-in-the-loop) keçirilməlidir.

## 📝 4. Markdown Formatlaşdırma və Sənədləşdirmə Qaydaları

Sənədləşdirmə və Notebook daxilində vizual oxunaqlılığı təmin etmək üçün aşağıdakı Markdown qaydalarına riayət olunmalıdır:

### 4.1. Başlıq İyerarxiyası (Headings)
- Hər bir sənəddə yalnız bir ədəd Ana Başlıq (`# H1`) olmalıdır.
- Bölmələr üçün `#`, `##`, `###` ardıcıllığı pozulmamalıdır.

### 4.2. Kod Blokları (Code Formatting)
- Kod nümunələri yazılarkən sintaksis işıqlandırılması üçün sintaksis dili mütləq göstərilməlidir (`python`, `bash`, `json`, `sql` və s.):
  ```python
  def hello_world():
      print("Hello, World!")
  ```

### 4.3. Mətn Fərqləndirməsi və Siyahılar
- Mühüm anlayışlar **qalın (bold)** və ya *kursiv (italic)* olaraq vurğulanmalıdır.
- Elementlər ardıcıllıq tələb edirsə nömrələnmiş (`1.`, `2.`), əks halda maddəli (`-` və ya `*`) siyahı istifadə olunmalıdır.

### 4.4. Cədvəllər (Tables)
- Strukturlaşdırılmış data və müqayisələr üçün Markdown cədvəl formatından istifadə olunmalıdır:
  | Parametr | Təsviri | Nümunə |
  | :--- | :--- | :--- |
  | `lr` | Öyrənmə dərəcəsi | `0.001` |
  | `epochs` | İterasiya sayı | `50` |

### 4.5. Qeydlər və Xəbərdarlıqlar (Quotes / Alerts)
- Xüsusi diqqət yetirilməli məqamlar üçün blok sitatlarından (`>`) istifadə edilməlidir:
  > 💡 **Məsləhət:** Notebook-u yadda saxlamadan öncə Output-ları təmizləmək tövsiyə olunur.
