```python
# Lazımi kitabxanaların idxalı
import pandas as pd
import numpy as np
```

## 📚 Tapşırıq 1: Kitablar Haqqında Məlumatlar

**Tələblər:**
1. `Title`, `Author`, `PublicationYear`, `Price` sütunlarından ibarət ən azı 6 sətirlik DataFrame yaradın.
2. Bütün DataFrame-i ekrana çıxarın.
3. Kitabların orta qiymətini hesablayın.
4. 2015-ci ildən sonra nəşr olunmuş kitabları tapın.
5. DataFrame-i `Price` sütununa görə artan sıra ilə çeşidləyin.

```python
# 1. DataFrame-in yaradılması (ən azı 6 sətir)
books_data = {
    'Title': ['1984', 'Səfillər', 'Xənsə', 'Qürur və Qərəz', 'Dahi', 'Şəfəq', 'Atom Ədədli Adətlər'],
    'Author': ['George Orwell', 'Victor Hugo', 'Nizami Gəncəvi', 'Jane Austen', 'Theodore Dreiser', 'Stephenie Meyer', 'James Clear'],
    'PublicationYear': [1949, 1862, 1188, 1813, 1915, 2005, 2018],
    'Price': [15.50, 24.00, 18.00, 12.50, 20.00, 16.80, 22.00]
}
df_books = pd.DataFrame(books_data)

print("=== 2. Bütün DataFrame ===")
display(df_books)

# 3. Kitabların orta qiyməti
avg_price = df_books['Price'].mean()
print(f"\n💰 Kitabların orta qiyməti: {avg_price:.2f} AZN")

# 4. 2015-ci ildən sonra nəşr olunmuş kitablar
after_2015 = df_books[df_books['PublicationYear'] > 2015]
print("\n📅 2015-ci ildən sonra nəşr olunmuş kitablar:")
display(after_2015)

# 5. Price sütununa görə artan sırada çeşidləmə
sorted_books = df_books.sort_values(by='Price', ascending=True)
print("\n📈 Qiymətə görə (artan sıra ilə) çeşidlənmiş DataFrame:")
display(sorted_books)
```

## 🛒 Tapşırıq 2: Sifarişlər (CSV-dən Oxuma)

**Tələblər:**
1. Məlumatları `orders.csv` faylından DataFrame-ə oxuyun (`OrderID`, `Customer`, `Date`, `Amount`).
2. İlk 10 sətri ekrana çıxarın.
3. Hər `Customer` üçün sifarişlərin sayını müəyyən edin.
4. Sifarişlərin maksimum və minimum `Amount` dəyərlərini tapın.
5. Bütün sifarişlərin ümumi məbləğini hesablayın.

```python
# 1. CSV faylından DataFrame-ə oxunması
df_orders = pd.read_csv('orders.csv')

# 2. İlk 10 sətrin ekrana çıxarılması
print("=== İlk 10 sətir ===")
display(df_orders.head(10))

# 3. Hər Customer üçün sifarişlərin sayı
customer_order_counts = df_orders['Customer'].value_counts()
print("\n👤 Hər müştəri üçün sifariş sayı:")
print(customer_order_counts)

# 4. Maksimum və minimum Amount dəyərləri
max_amount = df_orders['Amount'].max()
min_amount = df_orders['Amount'].min()
print(f"\n🔴 Maksimum Sifariş Məbləği: {max_amount} AZN")
print(f"🟢 Minimum Sifariş Məbləği: {min_amount} AZN")

# 5. Bütün sifarişlərin ümumi məbləği
total_amount = df_orders['Amount'].sum()
print(f"\n💵 Bütün sifarişlərin ümumi məbləği: {total_amount:.2f} AZN")
```

## 🍎 Tapşırıq 3: Qida Məhsulları

**Tələblər:**
1. `Product`, `Category`, `Calories`, `Protein` sütunlarından ibarət ən azı 10 sətirlik DataFrame yaradın.
2. Bütün DataFrame-i ekrana çıxarın.
3. `Calories` dəyəri 300-dən çox olan bütün məhsulları tapın.
4. `Category` üzrə orta `Protein` miqdarını hesablayın.
5. DataFrame-i `Calories` sütununa görə azalan sıra ilə çeşidləyin.

```python
# 1. DataFrame-in yaradılması (ən azı 10 sətir)
food_data = {
    'Product': ['Alma', 'Toyuq Qanadı', 'Qızılbalıq', 'Avokado', 'Yumurta', 'Süd', 'Şokolad', 'Badam', 'Banan', 'Qatıq'],
    'Category': ['Meyvə', 'Ət', 'Balıq', 'Meyvə', 'Zülal', 'Süd Məhsulu', 'Şirniyyat', 'Çərəz', 'Meyvə', 'Süd Məhsulu'],
    'Calories': [52, 220, 206, 160, 155, 62, 546, 579, 89, 59],
    'Protein': [0.3, 30.5, 22.0, 2.0, 13.0, 3.4, 4.9, 21.2, 1.1, 10.0]
}
df_food = pd.DataFrame(food_data)

print("=== 2. Bütün DataFrame ===")
display(df_food)

# 3. Calories > 300 olan məhsullar
high_cal = df_food[df_food['Calories'] > 300]
print("\n🔥 Kalorisi 300-dən çox olan məhsullar:")
display(high_cal)

# 4. Category üzrə orta Protein miqdarı
avg_protein = df_food.groupby('Category')['Protein'].mean()
print("\n🥩 Kategoriya üzrə orta Protein miqdarı (qramlıq):")
print(avg_protein)

# 5. Calories sütununa görə azalan sıra ilə çeşidləmə
sorted_food = df_food.sort_values(by='Calories', ascending=False)
print("\n📉 Kaloriyə görə (azalan sıra ilə) çeşidlənmiş DataFrame:")
display(sorted_food)
```

## 💼 Tapşırıq 4: Əməkdaşlar və Layihələr

**Tələblər:**
1. `Name`, `Project`, `Hours` sütunlarından ibarət ən azı 8 sətirlik DataFrame yaradın.
2. İlkin DataFrame-i ekrana çıxarın.
3. Hər `Name` üzrə ümumi iş saatını hesablayın.
4. Hər `Project` üzrə ümumi iş saatını hesablayın.
5. Ən çox saat işləyən əməkdaşı tapın.

```python
# 1. DataFrame-in yaradılması (ən azı 8 sətir)
emp_data = {
    'Name': ['Anar', 'Ləman', 'Anar', 'Rashad', 'Ləman', 'Rashad', 'Leyla', 'Anar'],
    'Project': ['Web Site', 'Mobile App', 'AI Bot', 'Mobile App', 'Web Site', 'AI Bot', 'Web Site', 'Mobile App'],
    'Hours': [40, 35, 25, 45, 20, 30, 50, 15]
}
df_emp = pd.DataFrame(emp_data)

print("=== 2. İlkin DataFrame ===")
display(df_emp)

# 3. Hər Name üzrə ümumi iş saatı
hours_per_person = df_emp.groupby('Name')['Hours'].sum()
print("\n⏱️ Hər əməkdaş üzrə ümumi iş saatı:")
print(hours_per_person)

# 4. Hər Project üzrə ümumi iş saatı
hours_per_project = df_emp.groupby('Project')['Hours'].sum()
print("\n📊 Hər layihə üzrə ümumi iş saatı:")
print(hours_per_project)

# 5. Ən çox saat işləyən əməkdaş
top_employee = hours_per_person.idxmax()
top_hours = hours_per_person.max()
print(f"\n🏆 Ən çox işləyən əməkdaş: {top_employee} ({top_hours} saat)")
```

## 🎟️ Tapşırıq 5: Bilet Satışları

**Tələblər:**
1. `Movie`, `City`, `TicketsSold` sütunlarından ibarət ən azı 12 sətirlik DataFrame yaradın.
2. Bütün DataFrame-i ekrana çıxarın.
3. Hər `Movie` üzrə satılmış biletlərin ümumi sayını hesablayın.
4. Hər `City` üzrə satılmış biletlərin ümumi sayını hesablayın.
5. Ən çox bilet satılan filmi tapın.

```python
# 1. DataFrame-in yaradılması (ən azı 12 sətir)
tickets_data = {
    'Movie': ['Avatar 2', 'Inception', 'Oppenheimer', 'Avatar 2', 'Oppenheimer', 'Inception', 
              'Interstellar', 'Avatar 2', 'Interstellar', 'Oppenheimer', 'Inception', 'Interstellar'],
    'City': ['Bakı', 'Bakı', 'Gəncə', 'Sumqayıt', 'Bakı', 'Gəncə', 
             'Bakı', 'Gəncə', 'Sumqayıt', 'Sumqayıt', 'Sumqayıt', 'Gəncə'],
    'TicketsSold': [450, 300, 200, 180, 520, 150, 380, 210, 190, 240, 280, 160]
}
df_tickets = pd.DataFrame(tickets_data)

print("=== 2. Bütün DataFrame ===")
display(df_tickets)

# 3. Hər Movie üzrə satılmış biletlərin ümumi sayı
tickets_per_movie = df_tickets.groupby('Movie')['TicketsSold'].sum()
print("\n🎬 Hər film üzrə ümumi bilet satışı:")
print(tickets_per_movie)

# 4. Hər City üzrə satılmış biletlərin ümumi sayı
tickets_per_city = df_tickets.groupby('City')['TicketsSold'].sum()
print("\n🏙️ Hər şəhər üzrə ümumi bilet satışı:")
print(tickets_per_city)

# 5. Ən çox bilet satılan film
top_movie = tickets_per_movie.idxmax()
top_movie_tickets = tickets_per_movie.max()
print(f"\n🍿 Ən çox bilet satılan film: '{top_movie}' ({top_movie_tickets} bilet)")
```
