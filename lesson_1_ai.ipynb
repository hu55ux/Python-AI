# Lesson 1: Süni İntellekt və Data Analizinə Giriş (Introduction to AI & Data Science)

Bu notebook **Lesson 1 AI** kursu çərçivəsində hazırlanıb. Burada siz **NumPy**, **Pandas** və **Matplotlib** kitabxanaları ilə ilkin tanışlıq əldə edəcəksiniz.

```python
# Kitabxanaların idxalı (Importing Libraries)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("NumPy Versiyası:", np.__version__)
print("Pandas Versiyası:", pd.__version__)
```

## 1. NumPy ilə Massivlərin Yaradılması

```python
# 1D və 2D NumPy Massivləri
data_1d = np.array([10, 20, 30, 40, 50])
data_2d = np.array([[1, 2, 3], [4, 5, 6]])

print("1D Array:", data_1d)
print("2D Array:\n", data_2d)
print("Orta qiymət (Mean):", np.mean(data_1d))
```

## 2. Pandas ilə Dataframe Yaradılması

```python
# Nümunə məlumatlar
df_data = {
    'Tələbə': ['Aysel', 'Elvin', 'Nigar', 'Murad'],
    'Bal': [85, 92, 78, 88],
    'Fənn': ['AI Basics', 'AI Basics', 'AI Basics', 'AI Basics']
}

df = pd.DataFrame(df_data)
df
```

## 3. Matplotlib ilə Qrafik Çəkilməsi

```python
plt.figure(figsize=(7, 4))
plt.bar(df['Tələbə'], df['Bal'], color=['#4F46E5', '#06B6D4', '#10B981', '#F59E0B'])
plt.title('Tələbələrin AI Dərsi Üzrə Nəticələri')
plt.xlabel('Tələbə')
plt.ylabel('Bal')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
