# Lesson 1: Data Analysis with Pandas

```python
import pandas as pd
import numpy as np
```

### Task 1: Books Dataset

```python
books_data = {
    'Title': ['1984', 'Les Misérables', 'The Great Gatsby', 'Pride and Prejudice', 'The Genius', 'Atomic Habits', 'Twilight'],
    'Author': ['George Orwell', 'Victor Hugo', 'F. Scott Fitzgerald', 'Jane Austen', 'Theodore Dreiser', 'James Clear', 'Stephenie Meyer'],
    'PublicationYear': [1949, 1862, 1925, 1813, 1915, 2018, 2005],
    'Price': [15.5, 24.0, 18.0, 12.5, 20.0, 22.0, 16.8]
}

df_books = pd.DataFrame(books_data)
print("Entire DataFrame:")
print(df_books)

avg_price = df_books['Price'].mean()
print("\nAverage book price:", round(avg_price, 2))

books_after_2015 = df_books[df_books['PublicationYear'] > 2015]
print("\nBooks published after 2015:")
print(books_after_2015)

sorted_books = df_books.sort_values(by='Price', ascending=True)
print("\nSorted by price (ascending):")
print(sorted_books)
```

### Task 2: Orders Dataset (orders.csv)

```python
df_orders = pd.read_csv('orders.csv')

print("First 10 rows:")
print(df_orders.head(10))

orders_per_customer = df_orders['Customer'].value_counts()
print("\nNumber of orders per customer:")
print(orders_per_customer)

max_amount = df_orders['Amount'].max()
min_amount = df_orders['Amount'].min()
print("\nMax order amount:", max_amount)
print("Min order amount:", min_amount)

total_amount = df_orders['Amount'].sum()
print("\nTotal order amount:", round(total_amount, 2))
```

### Task 3: Food Products Dataset

```python
food_data = {
    'Product': ['Apple', 'Chicken Breast', 'Salmon', 'Avocado', 'Egg', 'Milk', 'Chocolate', 'Almond', 'Banana', 'Yogurt'],
    'Category': ['Fruit', 'Meat', 'Fish', 'Fruit', 'Protein', 'Dairy', 'Sweets', 'Nuts', 'Fruit', 'Dairy'],
    'Calories': [52, 220, 206, 160, 155, 62, 546, 579, 89, 59],
    'Protein': [0.3, 30.5, 22.0, 2.0, 13.0, 3.4, 4.9, 21.2, 1.1, 10.0]
}

df_food = pd.DataFrame(food_data)
print("Entire DataFrame:")
print(df_food)

high_calories = df_food[df_food['Calories'] > 300]
print("\nProducts with calories > 300:")
print(high_calories)

avg_protein_by_cat = df_food.groupby('Category')['Protein'].mean()
print("\nAverage protein by category:")
print(avg_protein_by_cat)

sorted_food = df_food.sort_values(by='Calories', ascending=False)
print("\nSorted by calories (descending):")
print(sorted_food)
```

### Task 4: Employees and Projects Dataset

```python
emp_data = {
    'Name': ['Andrew', 'Sarah', 'Andrew', 'Richard', 'Sarah', 'Richard', 'Emily', 'Andrew'],
    'Project': ['Web Site', 'Mobile App', 'AI Bot', 'Mobile App', 'Web Site', 'AI Bot', 'Web Site', 'Mobile App'],
    'Hours': [40, 35, 25, 45, 20, 30, 50, 15]
}

df_emp = pd.DataFrame(emp_data)
print("Initial DataFrame:")
print(df_emp)

hours_by_name = df_emp.groupby('Name')['Hours'].sum()
print("\nTotal hours worked by employee:")
print(hours_by_name)

hours_by_project = df_emp.groupby('Project')['Hours'].sum()
print("\nTotal hours worked by project:")
print(hours_by_project)

max_worker = hours_by_name.idxmax()
max_hours = hours_by_name.max()
print(f"\nEmployee with maximum hours: {max_worker} ({max_hours} hours)")
```

### Task 5: Ticket Sales Dataset

```python
tickets_data = {
    'Movie': ['Avatar 2', 'Inception', 'Oppenheimer', 'Avatar 2', 'Oppenheimer', 'Inception', 
              'Interstellar', 'Avatar 2', 'Interstellar', 'Oppenheimer', 'Inception', 'Interstellar'],
    'City': ['New York', 'New York', 'London', 'Chicago', 'New York', 'London', 
             'New York', 'London', 'Chicago', 'Chicago', 'Chicago', 'London'],
    'TicketsSold': [450, 300, 200, 180, 520, 150, 380, 210, 190, 240, 280, 160]
}

df_tickets = pd.DataFrame(tickets_data)
print("Entire DataFrame:")
print(df_tickets)

tickets_by_movie = df_tickets.groupby('Movie')['TicketsSold'].sum()
print("\nTotal tickets sold per movie:")
print(tickets_by_movie)

tickets_by_city = df_tickets.groupby('City')['TicketsSold'].sum()
print("\nTotal tickets sold per city:")
print(tickets_by_city)

top_movie = tickets_by_movie.idxmax()
top_tickets = tickets_by_movie.max()
print(f"\nMovie with highest tickets sold: {top_movie} ({top_tickets} tickets)")
```
