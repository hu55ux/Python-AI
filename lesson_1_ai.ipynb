# Lesson 1: Pandas Ilə Data Analizi Tapşırıqları

```python
import pandas as pd
import numpy as np
```

### Tapşırıq 1: Kitablar haqqında məlumatlar

```python
# DataFrame yaradılması
data_books = {
    'Title': ['1984', 'Səfillər', 'Xəmsə', 'Qürur və Qərəz', 'Dahi', 'Atom Ədədli Adətlər', 'Şəfəq'],
    'Author': ['George Orwell', 'Victor Hugo', 'Nizami Gəncəvi', 'Jane Austen', 'Theodore Dreiser', 'James Clear', 'Stephenie Meyer'],
    'PublicationYear': [1949, 1862, 1188, 1813, 1915, 2018, 2005],
    'Price': [15.5, 24.0, 18.0, 12.5, 20.0, 22.0, 16.8]
}

df_books = pd.DataFrame(data_books)

# DataFrame-i ekrana çıxarmaq
print("Bütün DataFrame:")
print(df_books)

# Orta qiymətin hesablanması
avg_price = df_books['Price'].mean()
print("\nOrta qiymət:", round(avg_price, 2))

# 2015-ci ildən sonra nəşr olunanlar
books_after_2015 = df_books[df_books['PublicationYear'] > 2015]
print("\n2015-ci ildən sonra nəşr olunanlar:")
print(books_after_2015)

# Price sütununa görə artan sırada çeşidləmək
sorted_books = df_books.sort_values(by='Price', ascending=True)
print("\nQiymətə görə çeşidlənmiş (artan sıra):")
print(sorted_books)
```

### Tapşırıq 2: Sifarişlər (orders.csv)

```python
# CSV faylından oxumaq
df_orders = pd.read_csv('orders.csv')

# İlk 10 sətir
print("İlk 10 sətir:")
print(df_orders.head(10))

# Hər Customer üçün sifariş sayı
orders_per_customer = df_orders['Customer'].value_counts()
print("\nHər müştərinin sifariş sayı:")
print(orders_per_customer)

# Maksimum və minimum Amount
max_amount = df_orders['Amount'].max()
min_amount = df_orders['Amount'].min()
print("\nMaksimum məbləğ:", max_amount)
print("Minimum məbləğ:", min_amount)

# Ümumi məbləğ
total_amount = df_orders['Amount'].sum()
print("\nÜmumi sifariş məbləği:", round(total_amount, 2))
```

### Tapşırıq 3: Qida məhsulları

```python
# Məhsullar cədvəlinin yaradılması
data_food = {
    'Product': ['Alma', 'Toyuq əti', 'Qızılbalıq', 'Avokado', 'Yumurta', 'Süd', 'Şokolad', 'Badam', 'Banan', 'Qatıq'],
    'Category': ['Meyvə', 'Ət', 'Balıq', 'Meyvə', 'Zülal', 'Süd Məhsulu', 'Şirniyyat', 'Çərəz', 'Meyvə', 'Süd Məhsulu'],
    'Calories': [52, 220, 206, 160, 155, 62, 546, 579, 89, 59],
    'Protein': [0.3, 30.5, 22.0, 2.0, 13.0, 3.4, 4.9, 21.2, 1.1, 10.0]
}

df_food = pd.DataFrame(data_food)
print("Bütün DataFrame:")
print(df_food)

# Calories > 300 olanlar
high_calories = df_food[df_food['Calories'] > 300]
print("\nKalorisi 300-dən çox olan məhsullar:")
print(high_calories)

# Category üzrə orta Protein
avg_protein_by_cat = df_food.groupby('Category')['Protein'].mean()
print("\nKategoriyalar üzrə orta protein:")
print(avg_protein_by_cat)

# Calories-ə görə azalan sıra ilə çeşidləmək
sorted_food = df_food.sort_values(by='Calories', ascending=False)
print("\nKaloriyə görə azalan sırada:")
print(sorted_food)
```

### Tapşırıq 4: Əməkdaşlar və layihələr

```python
data_emp = {
    'Name': ['Anar', 'Ləman', 'Anar', 'Rəşad', 'Ləman', 'Rəşad', 'Leyla', 'Anar'],
    'Project': ['Web Site', 'Mobile App', 'AI Bot', 'Mobile App', 'Web Site', 'AI Bot', 'Web Site', 'Mobile App'],
    'Hours': [40, 35, 25, 45, 20, 30, 50, 15]
}

df_emp = pd.DataFrame(data_emp)
print("İlkin DataFrame:")
print(df_emp)

# Hər Name üzrə ümumi iş saatı
hours_by_name = df_emp.groupby('Name')['Hours'].sum()
print("\nHər əməkdaş üzrə ümumi saat:")
print(hours_by_name)

# Hər Project üzrə ümumi iş saatı
hours_by_project = df_emp.groupby('Project')['Hours'].sum()
print("\nHər layihə üzrə ümumi saat:")
print(hours_by_project)

# Ən çox işləyən əməkdaş
max_worker = hours_by_name.idxmax()
max_hours = hours_by_name.max()
print(f"\nƏn çox işləyən əməkdaş: {max_worker} ({max_hours} saat)")
```

### Tapşırıq 5: Bilet satışları

```python
data_tickets = {
    'Movie': ['Avatar 2', 'Inception', 'Oppenheimer', 'Avatar 2', 'Oppenheimer', 'Inception', 
              'Interstellar', 'Avatar 2', 'Interstellar', 'Oppenheimer', 'Inception', 'Interstellar'],
    'City': ['Bakı', 'Bakı', 'Gəncə', 'Sumqayıt', 'Bakı', 'Gəncə', 
             'Bakı', 'Gəncə', 'Sumqayıt', 'Sumqayıt', 'Sumqayıt', 'Gəncə'],
    'TicketsSold': [450, 300, 200, 180, 520, 150, 380, 210, 190, 240, 280, 160]
}

df_tickets = pd.DataFrame(data_tickets)
print("Bütün DataFrame:")
print(df_tickets)

# Hər Movie üzrə ümumi bilet satışı
tickets_by_movie = df_tickets.groupby('Movie')['TicketsSold'].sum()
print("\nHər film üzrə ümumi bilet satışı:")
print(tickets_by_movie)

# Hər City üzrə ümumi bilet satışı
tickets_by_city = df_tickets.groupby('City')['TicketsSold'].sum()
print("\nHər şəhər üzrə ümumi bilet satışı:")
print(tickets_by_city)

# Ən çox bilet satılan film
top_movie = tickets_by_movie.idxmax()
top_tickets = tickets_by_movie.max()
print(f"\nƏn çox bilet satılan film: {top_movie} ({top_tickets} bilet)")
```
