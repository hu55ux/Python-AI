# Lesson 2: Data Visualization with Matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np
```

### Homework 1: Weekly Temperature Line Plot

```python
# Homework 1: Weekly Temperature Line Plot
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temperature = [18, 20, 22, 21, 24, 26, 23]

plt.figure(figsize=(8, 5))
plt.plot(days, temperature, marker='o', linestyle='--', color='#2b5c8f', linewidth=2, markersize=8)

plt.title("Weekly Temperature Trend", fontsize=13)
plt.xlabel("Days", fontsize=11)
plt.ylabel("Temperature (C)", fontsize=11)
plt.grid(True, linestyle=':', alpha=0.7)

for i, temp in enumerate(temperature):
    plt.text(days[i], temp + 0.35, f"{temp}C", ha='center', va='bottom', fontsize=9, fontweight='bold')

plt.ylim(16, 28)
plt.tight_layout()
plt.show()
```

### Homework 2: Programming Languages Comparison

```python
# Homework 2: Programming Languages Comparison (Horizontal & Vertical Bar Charts)
languages = ["Python", "JavaScript", "Java", "C#", "C++"]
students = [35, 28, 22, 18, 15]

# Horizontal Bar Chart
plt.figure(figsize=(9, 5))
colors = ['#d9534f', '#337ab7', '#5bc0de', '#f0ad4e', '#5cb85c']

bars = plt.barh(languages, students, color=colors, height=0.55)

plt.title("Students per Programming Language", fontsize=13)
plt.xlabel("Number of Students", fontsize=11)
plt.ylabel("Programming Language", fontsize=11)
plt.gca().invert_yaxis()
plt.grid(axis='x', linestyle=':', alpha=0.7)

for bar, val in zip(bars, students):
    plt.text(val + 0.6, bar.get_y() + bar.get_height()/2, str(val), va='center', fontsize=10, fontweight='bold')

plt.xlim(0, 40)
plt.tight_layout()
plt.show()

# Additional: Vertical Bar Chart
plt.figure(figsize=(8, 5))
vbars = plt.bar(languages, students, color=colors, width=0.5)

plt.title("Students per Programming Language (Vertical)", fontsize=13)
plt.xlabel("Programming Language", fontsize=11)
plt.ylabel("Number of Students", fontsize=11)
plt.grid(axis='y', linestyle=':', alpha=0.7)

for bar, val in zip(vbars, students):
    plt.text(bar.get_x() + bar.get_width()/2, val + 0.7, str(val), ha='center', fontsize=10, fontweight='bold')

plt.ylim(0, 40)
plt.tight_layout()
plt.show()
```

### Homework 3: Student Results Analysis

```python
# Homework 3: Student Results Analysis (Scatter Plot & Inset Bar Chart)
study_hours = [1, 2, 2, 3, 4, 4, 5, 6, 7, 8]
scores = [45, 50, 55, 60, 65, 70, 74, 80, 88, 95]
ages = [18, 19, 18, 20, 21, 19, 22, 20, 23, 21]

fig = plt.figure(figsize=(10, 6))
ax = fig.add_axes([0.1, 0.1, 0.8, 0.8])

# Scatter plot
scatter = ax.scatter(study_hours, scores, c=ages, cmap='viridis', s=120, marker='o', edgecolors='black', alpha=0.9)

# Colorbar
cbar = fig.colorbar(scatter, ax=ax)
cbar.set_label('Age', fontsize=11)

ax.set_title("Study Hours vs Exam Score Analysis", fontsize=13)
ax.set_xlabel("Study Hours", fontsize=11)
ax.set_ylabel("Exam Score", fontsize=11)
ax.grid(True, linestyle=':', alpha=0.7)

# Highlight top score
max_idx = scores.index(max(scores))
ax.annotate(f'Top Score: {scores[max_idx]}', 
            xy=(study_hours[max_idx], scores[max_idx]), 
            xytext=(study_hours[max_idx] - 1.8, scores[max_idx] - 2),
            arrowprops=dict(facecolor='red', shrink=0.08, width=1.5, headwidth=7),
            fontsize=10, fontweight='bold', color='darkred')

# Additional: Inset bar chart for scores
ax_inset = fig.add_axes([0.18, 0.55, 0.28, 0.28])
ax_inset.bar(range(1, len(scores) + 1), scores, color='#5cb85c', alpha=0.75)
ax_inset.set_title("Scores Overview", fontsize=9)
ax_inset.set_xlabel("Student", fontsize=8)
ax_inset.set_ylabel("Score", fontsize=8)
ax_inset.tick_params(labelsize=8)

plt.show()
```
